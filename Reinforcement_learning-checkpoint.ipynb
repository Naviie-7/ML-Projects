```python
import gymnasium as gym
env = gym.make('CartPole-v1')
print("Success:", env.observation_space, env.action_space)
state, info=env.reset()
print(state)
```

```python
import torch
import torch.nn as nn

class QNetwork(nn.Module):
    def __init__(self, state_size=4, action_size=2):
        super().__init__()  
        self.fc1=nn.Linear(state_size, 64)
        self.fc2=nn.Linear(64, 64)
        self.fc3=nn.Linear(64, action_size)
    def forward(self, x):
        x=torch.relu(self.fc1(x))
        x=torch.relu(self.fc2(x))
        return self.fc3(x)       # raw Q-values, no activation on output
```

```python
import random
import numpy as np

def select_action(state, q_network, epsilon, action_size=2):
    if random.random()<epsilon:
        return random.randint(0, action_size-1)  #Explore
    else:
        with torch.no_grad():
            state_tensor=torch.FloatTensor(state).unsqueeze(0)
            q_values=q_network(state_tensor)
            return torch.argmax(q_values).item() #Exploit
```

```python
from collections import deque
import random

class ReplayBuffer:
    def __init__(self, capacity=10000):
        self.buffer=deque(maxlen=capacity)    # auto-discards oldest when full

    def push(self, state, action, reward, next_state, done):
        self.buffer.append((state, action, reward, next_state, done))

    def sample(self, batch_size):
        return random.sample(self.buffer, batch_size)

    def __len__(self):
        return len(self.buffer)
```

```python
import torch.optim as optim

env=gym.make('CartPole-v1')
q_network=QNetwork(state_size=4, action_size=2)
target_network = QNetwork(state_size=4, action_size=2)
target_network.load_state_dict(q_network.state_dict())
optimizer=optim.Adam(q_network.parameters(), lr=0.001)
replay_buffer=ReplayBuffer(capacity=10000)

epsilon=1.0
epsilon_min=0.01
epsilon_decay=0.995
gamma=0.99
batch_size=64
n_episodes=1000
target_update_freq = 10 

episode_rewards=[]

for episode in range(n_episodes):
    state,_=env.reset()
    total_reward=0
    done=False

    while not done:
        action=select_action(state, q_network, epsilon, action_size=2)
        next_state, reward, terminated, truncated,_=env.step(action)
        done=terminated or truncated

        replay_buffer.push(state, action, reward, next_state, done)
        state=next_state
        total_reward+=reward

        if len(replay_buffer) >=batch_size:
            batch=replay_buffer.sample(batch_size)
            states, actions, rewards, next_states, dones=zip(*batch)

            states=torch.FloatTensor(np.array(states))
            actions=torch.LongTensor(actions)
            rewards=torch.FloatTensor(rewards)
            next_states=torch.FloatTensor(np.array(next_states))
            dones=torch.FloatTensor(dones)

            current_q=q_network(states).gather(1, actions.unsqueeze(1)).squeeze()

            with torch.no_grad():
                next_q=target_network(next_states).max(1)[0]
                target_q=rewards+gamma*next_q*(1-dones)

            loss=nn.functional.mse_loss(current_q, target_q)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

    epsilon=max(epsilon_min, epsilon*epsilon_decay)
    episode_rewards.append(total_reward)

    if episode % target_update_freq == 0:
        target_network.load_state_dict(q_network.state_dict())


    if episode%20==0:
        print(f"Episode {episode}, Reward: {total_reward}, Epsilon: {epsilon:.3f}")
```

```python
import matplotlib.pyplot as plt

plt.plot(episode_rewards)
plt.xlabel('Episode')
plt.ylabel('Total Reward')
plt.title('DQN Training on CartPole')
plt.show()
```

```python
target_network = QNetwork(state_size=4, action_size=2)
target_network.load_state_dict(q_network.state_dict())  # copy weights

# in training loop, use target_network instead of q_network for next_q:
with torch.no_grad():
    next_q = target_network(next_states).max(1)[0]
    target_q = rewards + gamma * next_q * (1 - dones)
# periodically sync:
if episode % 10 == 0:
    target_network.load_state_dict(q_network.state_dict())
```
