```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
df=pd.read_csv('creditcard.csv')
print(df.shape)
print(df['Class'].value_counts())
print(df.isnull().sum().sum())
```

```python
normal_data=df[df['Class']==0]      #Legitimate transactions
anomaly_data=df[df['Class']==1]     #Fraud

# Training set: ONLY normal data — the model never sees fraud during "training"
X_train=normal_data.sample(frac=0.6, random_state=42).drop('Class', axis=1).values
remaining_normal=normal_data.drop(normal_data.sample(frac=0.6, random_state=42).index)

# Val/test: a MIX of normal + fraud, used only to pick a threshold and evaluate
X_val_normal=remaining_normal.sample(frac=0.5, random_state=42).drop('Class', axis=1).values
X_val_anomaly=anomaly_data.sample(frac=0.5, random_state=42).drop('Class', axis=1).values
y_val=np.concatenate([np.zeros(len(X_val_normal)), np.ones(len(X_val_anomaly))])
X_val=np.vstack([X_val_normal, X_val_anomaly])

X_test_normal = remaining_normal.drop(remaining_normal.sample(frac=0.5, random_state=42).index).drop('Class', axis=1).values
X_test_anomaly = anomaly_data.drop(anomaly_data.sample(frac=0.5, random_state=42).index).drop('Class', axis=1).values
y_test = np.concatenate([np.zeros(len(X_test_normal)), np.ones(len(X_test_anomaly))])
X_test = np.vstack([X_test_normal, X_test_anomaly])
```

```python
X_train[:, -1] = np.log(X_train[:, -1] + 1e-6)
X_val[:, -1] = np.log(X_val[:, -1] + 1e-6)
X_test[:, -1] = np.log(X_test[:, -1] + 1e-6)
def estimate_gaussian(X):
    mu=np.mean(X, axis=0)
    var=np.var(X, axis=0)
    return mu, var

mu, var=estimate_gaussian(X_train)

def gaussian_probability(X, mu, var):
    var=np.where(var==0, 1e-8, var)

    coefficient=1/np.sqrt(2*np.pi*var)
    exponent=-((X-mu)**2)/(2*var)
    probs=coefficient*np.exp(exponent)

    return np.prod(probs, axis=1)
```

```python
def select_threshold(y_val, p_val):
    best_epsilon, best_f1=0, 0
    p_val_nonzero = p_val[p_val > 0]
    log_min, log_max = np.log10(p_val_nonzero.min()), np.log10(p_val_nonzero.max())
    epsilons = np.logspace(log_min, log_max, 1000) 

    for epsilon in epsilons:
        preds=(p_val<epsilon).astype(int)

        tp = np.sum((preds == 1) & (y_val == 1))
        fp = np.sum((preds == 1) & (y_val == 0))
        fn = np.sum((preds == 0) & (y_val == 1))

        precision=tp/(tp+fp) if(tp+fp)>0 else 0
        recall=tp/(tp+fn) if(tp+fn)>0 else 0

        f1=2*precision*recall/(precision+recall) if(precision+recall)>0 else 0

        if f1>best_f1:
            best_f1=f1
            best_epsilon=epsilon

    return best_epsilon, best_f1

p_val = gaussian_probability(X_val, mu, var)
epsilon, f1 = select_threshold(y_val, p_val)
```

```python
mu, var = estimate_gaussian(X_train)
p_val = gaussian_probability(X_val, mu, var)
epsilon, f1 = select_threshold(y_val, p_val)
print("New epsilon:", epsilon, "New F1:", f1)
```

```python
p_test = gaussian_probability(X_test, mu, var)
test_preds = (p_test < epsilon).astype(int)

tp = np.sum((test_preds == 1) & (y_test == 1))
fp = np.sum((test_preds == 1) & (y_test == 0))
fn = np.sum((test_preds == 0) & (y_test == 1))
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1_test = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0

print(f"Precision: {precision:.3f}, Recall: {recall:.3f}, F1: {f1_test:.3f}")
```

```python
print("Fraud in test set:", int(y_test.sum()))
print("Flagged as fraud:", int(test_preds.sum()))
print("True positives:", tp, "| False positives:", fp)
```

```python
p_val_normal = gaussian_probability(X_val[y_val == 0], mu, var)
p_val_fraud = gaussian_probability(X_val[y_val == 1], mu, var)

print("Normal probs — min:", p_val_normal.min(), "max:", p_val_normal.max(), "median:", np.median(p_val_normal))
print("Fraud probs — min:", p_val_fraud.min(), "max:", p_val_fraud.max(), "median:", np.median(p_val_fraud))
```

```python
# plot the full precision-recall tradeoff curve to see the whole picture
precisions, recalls = [], []
epsilons = np.logspace(np.log10(p_val[p_val>0].min()), np.log10(p_val.max()), 1000)

for epsilon in epsilons:
    preds = (p_val < epsilon).astype(int)
    tp = np.sum((preds == 1) & (y_val == 1))
    fp = np.sum((preds == 1) & (y_val == 0))
    fn = np.sum((preds == 0) & (y_val == 1))
    precisions.append(tp / (tp + fp) if (tp + fp) > 0 else 0)
    recalls.append(tp / (tp + fn) if (tp + fn) > 0 else 0)

plt.plot(recalls, precisions)
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Tradeoff (log-spaced thresholds)')
plt.show()
```

```python
best_idx=np.argmax([2*p*r/(p+r) if (p+r)>0 else 0 for p, r in zip(precisions, recalls)])
plt.plot(recalls, precisions)
plt.scatter(recalls[best_idx], precisions[best_idx], color='red', zorder=5, label=f'F1-optimal (F1={f1:.3f})')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Tradeoff (log-spaced thresholds)')
plt.legend()
plt.show()
```
